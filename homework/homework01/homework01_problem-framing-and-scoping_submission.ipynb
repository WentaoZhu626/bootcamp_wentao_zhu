## Problem Statement

Portfolio risk managers must decide whether to maintain, reduce, or hedge market exposure before the next trading session, but the size of the next day's market move is unknown when that decision is made. Relying only on fixed risk limits or a long historical average can understate risk when conditions change quickly. This project will investigate whether recent SPY returns, trading ranges, volume, and the VIX can provide a more responsive estimate of next-day market volatility.

The initial target will be the next trading day's absolute close-to-close SPY log return, used as a transparent daily volatility proxy. The forecast will use only information available by the current market close and will support a human review rather than automate a trade. Success is defined as at least a 5% reduction in holdout MAE relative to a 20-day rolling baseline, with a secondary alert target of at least 60% recall and 40% precision for high-volatility days.

## Stakeholder & User

- **Primary stakeholder:** Head of Portfolio Risk, who owns risk limits and escalation standards.
- **End users:** Daily risk analysts and portfolio managers who interpret the signal.
- **Decision:** Maintain exposure, reduce position sizes, or add hedges after reviewing a High risk signal.
- **Timing:** After the U.S. market close and before the next trading session.
- **User priorities:** Timeliness, stability, interpretability, clear uncertainty, and useful warnings during stressed periods.

## Useful Answer & Decision

This is a **predictive** problem. The main artifact will contain a next-day volatility estimate, a transparent rolling baseline, and a Low / Normal / High risk label. A High label triggers review of exposure, concentration, and hedge coverage; it does not automatically execute a trade. Primary evaluation uses chronological-holdout MAE, with RMSE and high-alert precision and recall as secondary diagnostics.

## Assumptions & Constraints

- Daily adjusted SPY data and VIX data are available and can be aligned by trading date.
- Predictors must be available by the current market close.
- The first version uses public daily data and open-source Python on a personal laptop.
- Intraday feeds, proprietary positions, options surfaces, and automated execution are out of scope.
- Raw data remains unchanged and processed outputs are reproducible from code.
- Evaluation uses chronological splits or walk-forward validation.
- Results are academic decision support, not investment advice.

## Known Unknowns / Risks

- **Regime change:** Use walk-forward tests and report calm and stressed periods separately.
- **Rare shocks:** Report tail-period errors and alert recall, not only average accuracy.
- **Leakage:** Lag predictors and fit transformations and thresholds on training data only.
- **Data quality:** Validate timestamps, missing dates, and adjustments and retain provenance.
- **Weak incremental signal:** Compare every model with a simple baseline and accept a negative result.
- **Alert fatigue:** Report precision-recall tradeoffs and document the threshold choice.
